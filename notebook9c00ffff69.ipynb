```python
train_data = pd.read_csv('/kaggle/input/competitions/titanic/train.csv')
train_data.shape
```

```python
train_data.isnull().sum()
```

```python
train_data.groupby('Sex')['Survived'].mean()
```

```python
train_data.groupby('Pclass')['Survived'].mean()
```

```python
train_data['Age']= train_data['Age'].fillna(train_data['Age'].median())
train_data['Sex']= train_data['Sex'].map({'female':0, 'male':1})
train_data['Embarked']= train_data['Embarked'].fillna('S')
train_data['Embarked']= train_data['Embarked'].map({'S':0, 'C':1, 'Q':2})
train_data.isnull().sum()
```

```python
train_data.isnull().sum()
```

```python
from sklearn.ensemble import RandomForestClassifier
features= ['Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare', 'Embarked']
x=train_data[features]
y=train_data['Survived']
model = RandomForestClassifier(n_estimators=100, max_depth=5, random_state=1)
model.fit(x,y)
```

```python
test_data = pd.read_csv('/kaggle/input/competitions/titanic/test.csv')
test_data.shape
```

```python
test_data.isnull().sum()
```

```python
test_data['Age'] = test_data['Age'].fillna(test_data['Age'].median())
test_data['Fare'] = test_data['Fare'].fillna(test_data['Fare'].median())
test_data['Emabarked']= test_data['Embarked'].fillna('S')
test_data['Sex']= test_data['Sex'].replace({'female':0, 'male':1})
test_data['Embarked']= test_data['Embarked'].replace({'S':0, 'C':1, 'Q':2})
x_test = test_data[features]
predictions= model.predict(x_test)
output = pd.DataFrame({'PassangerId': test_data.PassengerId, 'Survived': predictions})
output.to_csv('submission.csv', index=False)
print('Arquivo csv gerado com sucesso')
```
